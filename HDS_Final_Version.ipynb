```python
import pandas as pd
from sklearn.metrics import accuracy_score, roc_auc_score, precision_score, recall_score, f1_score

# Initialize result list
performance_results_rf = []

# Helper function to add RF model results
def add_result_rf(name, y_true, y_pred, y_prob, train_acc):
    performance_results_rf.append({
        "Model": name,
        "Train Accuracy": train_acc,
        "Test Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred),
        "Recall": recall_score(y_true, y_pred),
        "F1-Score": f1_score(y_true, y_pred),
        "ROC-AUC": roc_auc_score(y_true, y_prob)
    })

# -----------------------------------
# ADD RESULTS FOR RANDOM FOREST
# -----------------------------------

# 1. No Sampling
add_result_rf(
    "No Sampling",
    y_test,
    y_test_pred_rf,
    y_test_prob_rf,
    train_acc_rf
)

# 2. SMOTE
add_result_rf(
    "SMOTE",
    y_test,
    y_test_pred_smote,
    y_test_prob_smote,
    train_acc_smote # Corrected from train_acc_rf_smote
)

# 3. SMOTEENN
add_result_rf(
    "SMOTEENN",
    y_test,
    y_test_pred_smote_enn,
    y_test_prob_smote_enn,
    train_acc_smote_enn # Corrected from train_acc_rf_smote_enn
)

# 4. ADASYN
add_result_rf(
    "ADASYN",
    y_test,
    y_test_pred_ada,
    y_test_prob_ada,
    train_acc_ada # Corrected from train_acc_rf_ada
)

# 5. Hybrid (ADASYN→SMOTE)
add_result_rf(
    "Hybrid (ADASYN→SMOTE)",
    y_test,
    y_test_pred_hybrid,
    y_test_prob_hybrid,
    train_acc_hybrid
)

# -----------------------------------
# CREATE DATAFRAME
# -----------------------------------
performance_df_rf = pd.DataFrame(performance_results_rf)

# Sort by ROC-AUC descending
performance_df_rf = performance_df_rf.sort_values(by="ROC-AUC", ascending=False).reset_index(drop=True)

# Display results
print(performance_df_rf)
```
